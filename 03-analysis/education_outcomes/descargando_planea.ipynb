```python
import unicodedata
import re

def cargar_y_limpiar_siged(ruta_csv):
    # 1. Leer el archivo manejando posibles bytes distorsionados
    with open(ruta_csv, 'r', encoding='utf-8', errors='replace') as f:
        df = pd.read_csv(f)

    # 2. Diccionario de reparación de nombres de columnas
    # Las columnas clave son la 0 (cct) y las últimas dos (longitud y latitud)
    nombres_limpios = {
        df.columns[0]: 'cct',
        df.columns[1]: 'cve_turno',
        df.columns[2]: 'turno',
        df.columns[3]: 'nombre_escuela',
        df.columns[4]: 'tipo_educativo',
        df.columns[5]: 'nivel_educativo',
        df.columns[6]: 'servicio_educativo',
        df.columns[7]: 'control',
        df.columns[8]: 'sostenimiento',
        df.columns[9]: 'cve_entidad',
        df.columns[10]: 'entidad',
        df.columns[11]: 'cve_municipio',
        df.columns[12]: 'municipio',
        df.columns[13]: 'cve_localidad',
        df.columns[14]: 'localidad',
        df.columns[15]: 'domicilio',
        df.columns[16]: 'num_exterior',
        df.columns[20]: 'cve_colonia',
        df.columns[21]: 'colonia',
        df.columns[22]: 'codigo_postal',
        df.columns[35]: 'longitud',
        df.columns[36]: 'latitud'
    }
    
    # Renombrar las columnas mapeadas
    df = df.rename(columns=nombres_limpios)

    # 3. Diccionario de reemplazo para reparar los caracteres corruptos en el texto
    reemplazos = {
        'MɘICO': 'MÉXICO',
        'CUAUHTɍOC': 'CUAUHTÉMOC',
        'PڂLICO': 'PÚBLICO',
        'BSICA': 'BÁSICA',
        'TɃNICA': 'TÉCNICA',
        'V̓TOR': 'VÍCTOR',
        'COYOACN': 'COYOACÁN',
        'MAŔ': 'MARÍA',
        'CRESTӎ': 'CRESTÓN',
        '\ufeff': ''
    }

    # Aplicar corrección a todas las columnas de texto
    for col in df.select_dtypes(include='object').columns:
        for roto, arreglado in reemplazos.items():
            df[col] = df[col].astype(str).str.replace(roto, arreglado, regex=False)
        # Limpiar espacios en blanco al inicio y al final
        df[col] = df[col].str.strip()


    # Corrección complementaria para municipios y nombres
    reemplazos_extra = {
        'ɠ': 'É',
        'ӎ': 'ÓN',
        'ɘ': 'É',
        'ɍ': 'É',
        'ڂ': 'Ú',
        '̓': 'Í',
        '́': 'Í'
    }

    for col in ['municipio', 'localidad', 'nombre_escuela']:
        if col in df_siged_limpio.columns:
            for roto, arreglado in reemplazos_extra.items():
                df_siged_limpio[col] = df_siged_limpio[col].str.replace(roto, arreglado, regex=False)


    # Asegurar formato estándar de CCT y coordenadas numéricas
    df['cct'] = df['cct'].str.upper()
    df['longitud'] = pd.to_numeric(df['longitud'], errors='coerce')
    df['latitud'] = pd.to_numeric(df['latitud'], errors='coerce')

    return df

# Ejecutar la limpieza
df_siged_limpio = cargar_y_limpiar_siged('../../02-data/raw/siged/concentrado_escuelas_secundarias_zmvm.csv')

# Verificar columnas y muestra limpia
print("Columnas disponibles:", df_siged_limpio.columns.tolist()[:15])
display(df_siged_limpio[['cct', 'turno', 'nombre_escuela', 'municipio', 'longitud', 'latitud']].head())
```

```python
import geopandas as gpd

# 1. Cruzar las 4,683 CCTs balanceadas con el catálogo de SIGED
# Quedarse con un registro geográfico único por inmueble (CCT)
sec_balanceadas = df_siged_limpio[df_siged_limpio['cct'].isin(ccts_balanceadas)].copy()
sec_balanceadas = sec_balanceadas.drop_duplicates(subset=['cct']).reset_index(drop=True)

print(f"Secundarias balanceadas encontradas en SIGED: {len(sec_balanceadas)} de {len(ccts_balanceadas)}")

# 2. Descartar filas sin coordenadas válidas
sec_balanceadas = sec_balanceadas.dropna(subset=['latitud', 'longitud']).copy()

# 3. Proyectar a UTM Zona 14N (EPSG:32614)
gdf_rec = gpd.GeoDataFrame(
    sec_balanceadas,
    geometry=gpd.points_from_xy(sec_balanceadas['longitud'], sec_balanceadas['latitud']),
    crs="EPSG:4326"
).to_crs("EPSG:32614")

# 4. Delimitar estrictamente al dominio de los mosaicos SRTM (19°N a 20°N, 98°W a 100°W)
gdf_rec_zmvm = gdf_rec[
    (gdf_rec['latitud'] >= 19.005) & (gdf_rec['latitud'] <= 19.995) &
    (gdf_rec['longitud'] >= -99.995) & (gdf_rec['longitud'] <= -98.005)
].copy().reset_index(drop=True)

gdf_rec_zmvm['X_UTM'] = gdf_rec_zmvm.geometry.x.round(2)
gdf_rec_zmvm['Y_UTM'] = gdf_rec_zmvm.geometry.y.round(2)
gdf_rec_zmvm['RECID'] = [f"R{i+1:05d}" for i in range(len(gdf_rec_zmvm))]

# 5. Exportar el catálogo maestro para los análisis econométricos posteriores
gdf_rec_zmvm[['RECID', 'cct', 'X_UTM', 'Y_UTM', 'latitud', 'longitud']].to_csv(
    'catalogo_receptores_secundarias_zmvm.csv', index=False
)

# 6. Escribir el bloque de receptores para AERMAP
with open('bloque_re_aermap.txt', 'w') as f:
    f.write("RE STARTING\n")
    for _, row in gdf_rec_zmvm.iterrows():
        f.write(f"   DISCCART  {row['X_UTM']:11.2f}  {row['Y_UTM']:11.2f}  0.0  0.0\n")
    f.write("RE FINISHED\n")

print(f"Listo: {len(gdf_rec_zmvm)} receptores listos en 'bloque_re_aermap.txt'")
```

```python
# 1. Obtener los puntos físicos únicos (redondeados a centímetros/metros)
gdf_rec_unicos = gdf_rec_zmvm.drop_duplicates(subset=['X_UTM', 'Y_UTM']).copy().reset_index(drop=True)

# 2. Asignar el identificador limpio de receptor
gdf_rec_unicos['RECID'] = [f"R{i+1:05d}" for i in range(len(gdf_rec_unicos))]

# 3. Asignar el RECID de vuelta a todas las CCTs (crear la tabla puente)
df_puente_receptores = gdf_rec_zmvm[['cct', 'turno', 'X_UTM', 'Y_UTM']].merge(
    gdf_rec_unicos[['X_UTM', 'Y_UTM', 'RECID']],
    on=['X_UTM', 'Y_UTM'],
    how='left'
)

# Guardar la tabla relacional para la econometría
df_puente_receptores.to_csv('tabla_enlace_cct_recid.csv', index=False)

# 4. Generar el bloque RE definitivo para AERMAP solo con puntos únicos
with open('bloque_re_aermap.txt', 'w') as f:
    f.write("RE STARTING\n")
    for _, row in gdf_rec_unicos.iterrows():
        f.write(f"   DISCCART  {row['X_UTM']:11.2f}  {row['Y_UTM']:11.2f}  0.0  0.0\n")
    f.write("RE FINISHED\n")

print(f"Total CCTs evaluadas: {len(gdf_rec_zmvm)}")
print(f"Inmuebles / Receptores físicos únicos para AERMAP: {len(gdf_rec_unicos)}")
```
